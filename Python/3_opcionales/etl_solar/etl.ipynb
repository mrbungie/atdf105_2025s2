# ETL Solar Completo

Pipeline ETL con multiples fuentes de plantas solares.

Conversion conceptual 1:1 desde el ejemplo R homologo.


```python
from pathlib import Path

NOTEBOOK_DIR = Path.cwd()


def find_data(filename: str) -> Path:
    candidates = [NOTEBOOK_DIR / filename, NOTEBOOK_DIR / "data" / filename]
    for parent in [NOTEBOOK_DIR, *NOTEBOOK_DIR.parents]:
        candidates.append(parent / filename)
        candidates.append(parent / "data" / filename)
    for candidate in candidates:
        if candidate.exists():
            return candidate
    raise FileNotFoundError(f"No se encontro el archivo de datos: {filename}")

print("Notebook dir:", NOTEBOOK_DIR)
```

```python
import pandas as pd

raw_dir = NOTEBOOK_DIR / "raw_data"
files = sorted(raw_dir.glob("*.csv"))
print("Archivos fuente:", [f.name for f in files])

frames = {}
for f in files:
    frames[f.stem] = pd.read_csv(f)
    print(f.stem, frames[f.stem].shape)
```

```python
for name, frame in frames.items():
    frame.columns = [c.strip().lower() for c in frame.columns]

summary = {k: v.isna().mean().sort_values(ascending=False).head(5) for k, v in frames.items()}
for k, v in summary.items():
    print("
", k)
    print(v)
```
