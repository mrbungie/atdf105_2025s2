# Carga y Tests en Adult

Carga del dataset Adult y pruebas/estadistica descriptiva inicial.

Conversion conceptual 1:1 desde el ejemplo R homologo.


```python
from pathlib import Path


def find_repo_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / "R").exists():
            return candidate
    raise FileNotFoundError("No se encontro la carpeta R del repositorio")

REPO_ROOT = find_repo_root(Path.cwd())
print("Repo root:", REPO_ROOT)
```

```python
import pandas as pd
from scipy import stats

adult_path = REPO_ROOT / "R" / "nuevos" / "1_inicio_adult" / "adult.csv"
df = pd.read_csv(adult_path)
print(df.head())
print(df.describe(include="all").transpose().head(12))
```

```python
# Ejemplo de contraste simple: diferencia de horas por nivel de ingreso
x = df.loc[df["income_status"] == "<=50K", "hours_per_week"].dropna()
y = df.loc[df["income_status"] == ">50K", "hours_per_week"].dropna()
stat, pvalue = stats.ttest_ind(x, y, equal_var=False)
print({"t_stat": stat, "pvalue": pvalue})

# Tabla de frecuencias de ingreso
print(df["income_status"].value_counts(normalize=True))
```
