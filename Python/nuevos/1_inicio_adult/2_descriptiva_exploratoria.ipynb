# EDA Adult

Analisis descriptivo y visual exploratorio del dataset Adult.

Conversion conceptual 1:1 desde el ejemplo R homologo.


```python
from pathlib import Path


def find_repo_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / "R").exists():
            return candidate
    raise FileNotFoundError("No se encontro la carpeta R del repositorio")

REPO_ROOT = find_repo_root(Path.cwd())
print("Repo root:", REPO_ROOT)
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

adult_path = REPO_ROOT / "R" / "nuevos" / "1_inicio_adult" / "adult.csv"
df = pd.read_csv(adult_path)

print(df.shape)
print(df.isna().mean().sort_values(ascending=False).head(10))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.countplot(data=df, x="income_status", ax=axes[0])
axes[0].set_title("Distribucion de ingreso")

sns.boxplot(data=df, x="income_status", y="hours_per_week", ax=axes[1])
axes[1].set_title("Horas por ingreso")
plt.tight_layout()
```
