# Limpieza Bank

Limpieza y transformacion del dataset de marketing bancario.

Conversion conceptual 1:1 desde el ejemplo R homologo.


```python
from pathlib import Path


def find_repo_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / "R").exists():
            return candidate
    raise FileNotFoundError("No se encontro la carpeta R del repositorio")

REPO_ROOT = find_repo_root(Path.cwd())
print("Repo root:", REPO_ROOT)
```

```python
import pandas as pd
import numpy as np

bank_path = REPO_ROOT / "R" / "nuevos" / "2_limpieza_bank" / "bank.csv"
df = pd.read_csv(bank_path, sep=';')

for col in df.columns:
    if df[col].dtype == object:
        df[col] = df[col].replace("", np.nan)

print(df.isna().sum().sort_values(ascending=False).head(10))
```

```python
df["balance_per_campaign"] = df["balance"] / df["campaign"].replace(0, np.nan)
df["has_previous_contact"] = (df["previous"] > 0).astype(int)
clean = df.drop_duplicates().copy()
print(clean.shape)
print(clean.head())
```
