# Red Bayesiana Titanic

Clasificacion con red bayesiana (aprox. con pgmpy) en Titanic.

Conversion conceptual 1:1 desde el ejemplo R homologo.


```python
from pathlib import Path


def find_repo_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / "R").exists():
            return candidate
    raise FileNotFoundError("No se encontro la carpeta R del repositorio")

REPO_ROOT = find_repo_root(Path.cwd())
print("Repo root:", REPO_ROOT)
```

```python
import pandas as pd

data_path = REPO_ROOT / "R" / "nuevos" / "5_aprendizaje_supervisado" / "data" / "titanic.csv"
df = pd.read_csv(data_path)

cols = ["Survived", "Pclass", "Sex", "Embarked", "SibSp", "Parch"]
work = df[cols].copy().fillna(df[cols].mode().iloc[0])
for c in ["Pclass", "SibSp", "Parch"]:
    work[c] = work[c].astype(str)

print(work.head())
print("Listo para pgmpy (equivalente a bnlearn)")
```
