# Clustering Jerarquico Iris

Clusterizacion jerarquica (aglomerativa/divisiva conceptual) en iris.

Conversion conceptual 1:1 desde el ejemplo R homologo.


```python
from pathlib import Path


def find_repo_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / "R").exists():
            return candidate
    raise FileNotFoundError("No se encontro la carpeta R del repositorio")

REPO_ROOT = find_repo_root(Path.cwd())
print("Repo root:", REPO_ROOT)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import linkage, dendrogram, fcluster
from sklearn.preprocessing import StandardScaler

iris_path = REPO_ROOT / "R" / "nuevos" / "6_aprendizaje_no_supervisado" / "data" / "iris.csv"
df = pd.read_csv(iris_path)
X = df.drop(columns=["variety"])
X_scaled = StandardScaler().fit_transform(X)

Z = linkage(X_scaled, method="ward")
plt.figure(figsize=(10, 4))
dendrogram(Z, truncate_mode="lastp", p=20)
plt.title("Dendrograma (Ward)")
plt.show()
```

```python
clusters = fcluster(Z, t=3, criterion="maxclust")
df["cluster_h"] = clusters
print(df[["variety", "cluster_h"]].head())
print(df.groupby(["cluster_h", "variety"]).size())
```
