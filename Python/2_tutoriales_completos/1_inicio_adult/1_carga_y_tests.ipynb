# Carga y Tests en Adult

Carga del dataset Adult y pruebas/estadistica descriptiva inicial.

```python
import pandas as pd
from scipy import stats

df = pd.read_csv("adult.csv", na_values="?")  # Cargar datos, marcando "?" como faltante

df.head()
```

```python
df.describe(include="all").transpose().head(12)
```

```python
# Ejemplo de contraste simple: diferencia de horas por nivel de ingreso
x = df.loc[df["income_status"] == "<=50K", "hours-per-week"].dropna()
y = df.loc[df["income_status"] == ">50K", "hours-per-week"].dropna()
```

```python
stat, pvalue = stats.ttest_ind(x, y, equal_var=False)
print({"t_stat": stat, "pvalue": pvalue})
```
