# ACP Original

Analisis de componentes principales (ACP/PCA) sobre planilla Excel original.

Conversion conceptual 1:1 desde el ejemplo R homologo.


```python
from pathlib import Path


def find_repo_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / "R").exists():
            return candidate
    raise FileNotFoundError("No se encontro la carpeta R del repositorio")

REPO_ROOT = find_repo_root(Path.cwd())
print("Repo root:", REPO_ROOT)
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

xlsx_path = REPO_ROOT / "R" / "originales" / "2_ACP" / "BBDD_EPA_NIVELES_ABSTRACCION_CLASS.xlsx"
df = pd.read_excel(xlsx_path)

num_df = df.select_dtypes(include=["number"]).dropna()
X_scaled = StandardScaler().fit_transform(num_df)

pca = PCA()
pca.fit(X_scaled)
print("Varianza explicada:", pca.explained_variance_ratio_[:10])
print("Acumulada:", pca.explained_variance_ratio_.cumsum()[:10])
```
