Estadistica descriptiva y visualizacion base sobre planilla Excel original.

Conversion conceptual 1:1 desde el ejemplo R homologo.


```python
from pathlib import Path


def find_repo_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / "R").exists():
            return candidate
    raise FileNotFoundError("No se encontro la carpeta R del repositorio")

REPO_ROOT = find_repo_root(Path.cwd())
print("Repo root:", REPO_ROOT)
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

xlsx_path = REPO_ROOT / "R" / "originales" / "1_Primer_Ejemplo_R" / "BBDD_EPA_NIVELES_ABSTRACCION_CLASS.xlsx"
df = pd.read_excel(xlsx_path)
print(df.head())
print(df.describe(include="all").transpose().head(15))
```

```python
num_cols = df.select_dtypes(include=["number"]).columns[:4]
if len(num_cols) >= 1:
    sns.histplot(df[num_cols[0]].dropna(), kde=True)
    plt.title(f"Distribucion de {num_cols[0]}")
    plt.show()
```
