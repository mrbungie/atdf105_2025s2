# ETL Solar Completo

Pipeline ETL con multiples fuentes de plantas solares.

Conversion conceptual 1:1 desde el ejemplo R homologo.


```python
from pathlib import Path


def find_repo_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / "R").exists():
            return candidate
    raise FileNotFoundError("No se encontro la carpeta R del repositorio")

REPO_ROOT = find_repo_root(Path.cwd())
print("Repo root:", REPO_ROOT)
```

```python
import pandas as pd

raw_dir = REPO_ROOT / "R" / "opcionales" / "etl_solar" / "raw_data"
files = sorted(raw_dir.glob("*.csv"))
print("Archivos fuente:", [f.name for f in files])

frames = {}
for f in files:
    frames[f.stem] = pd.read_csv(f)
    print(f.stem, frames[f.stem].shape)
```

```python
for name, frame in frames.items():
    frame.columns = [c.strip().lower() for c in frame.columns]

summary = {k: v.isna().mean().sort_values(ascending=False).head(5) for k, v in frames.items()}
for k, v in summary.items():
    print("
", k)
    print(v)
```
