## 3) Estadisticos de Edad

```python
# Limpiar columna numerica
edad = pd.to_numeric(data["Edad"], errors="coerce")
```

```python
# Eliminar valores faltantes
edad = edad.dropna()
```

```python
# Media
edad.mean()
```

```python
# Mediana
edad.median()
```

```python
# Minimo
edad.min()
```

```python
# Maximo
edad.max()
```

```python
# Rango
edad.max() - edad.min()
```

```python
# Varianza
edad.var()
```

```python
# Desviacion estandar
edad.std()
```

```python
# Asimetria
skew(edad)
```

```python
# Curtosis
kurtosis(edad)
```

## 5) Tablas de frecuencia

```python
# Tabla por modalidad
tabla_modalidad = data["Modalidad"].value_counts(dropna=False)
```

```python
# Ver tabla
tabla_modalidad
```

```python
# Para hacer gráficos más complejos necesitamos 
# poner todo en el mismo lugar

# Grafico de barras
tabla_modalidad.plot(kind="bar")
# Etiquetas
plt.title("Estudiantes segun Jornada")
# Etiquetas
plt.xlabel("Jornada")
plt.ylabel("Frecuencia")
plt.show()
```

```python
# Mostrar
plt.show()
```

```python
# Tabla por semestre
tabla_semestre = data["Semestre"].value_counts(dropna=False).sort_index()
```

```python
# Ver tabla
tabla_semestre
```

```python
# Grafico de barras
tabla_semestre.plot(kind="bar")
plt.title("Estudiantes segun Semestre")
plt.xlabel("Semestre")
plt.ylabel("Frecuencia")
plt.show()
```

```python
# Tabla por abstraccion
tabla_abstraccion = data["Abstraccion final"].value_counts(dropna=False)
```

```python
# Ver tabla
tabla_abstraccion
```

```python
# Grafico de barras
tabla_abstraccion.plot(kind="bar")
plt.title("Abstraccion final")
plt.xlabel("Tipo de abstraccion")
plt.ylabel("Frecuencia")
plt.show()
```

## 6) Correlacion simple

```python
# Seleccionar columnas numericas de interes
corr = data[["Edad", "Semestre", "Pregunta1", "Pregunta2"]].copy()
```

```python
# Convertir a numerico
corr = corr.apply(pd.to_numeric, errors="coerce")
```

```python
# Eliminar faltantes
corr = corr.dropna()
```

```python
# Matriz de correlacion
correlacion = corr.corr().round(2)
```

```python
# Ver matriz
correlacion
```

```python
# Heatmap
plt.imshow(correlacion, cmap="coolwarm", vmin=-1, vmax=1)
# Ticks eje X
plt.xticks(range(len(correlacion.columns)), correlacion.columns, rotation=45)
# Titulo
plt.title("Correlacion de las variables")
# Ticks eje Y
plt.yticks(range(len(correlacion.index)), correlacion.index)
# Barra de color
plt.colorbar()
# Mostrar
plt.show()
```
