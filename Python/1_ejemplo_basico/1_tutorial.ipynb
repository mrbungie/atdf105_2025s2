```python
# Tipos basicos y operaciones
entero = 10
decimal = 3.14
texto = "Hola Python"
booleano = True

print(type(entero), type(decimal), type(texto), type(booleano))
print("Potencia:", 2 ** 3)
print("Raiz:", 16 ** 0.5)
```

```python
# Estructuras equivalentes a vectores/listas/data frames
import pandas as pd

# Podemos hacer listas/vectores (lista tiene varios tipos, vector uno solo)
vector = [1, 2, 3, 4, 5]
lista = ["a", 10, 2.5, False]

# Podemos hacer dataframes
df = pd.DataFrame({"x": [1, 2, 3], "y": [10, 20, 30]})
print("Vector:", vector)
print("Lista:", lista)
print(df)
```
